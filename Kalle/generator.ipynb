```python
import os
from sdv.utils import load_synthesizer

# Load your already-trained synthesizer
synthesizer = load_synthesizer(filepath="./.snapshots/par_synthesizer_windowed.pkl")

# Generate ~100,000 rows (667 windows x 150 rows)
synthetic_data = synthesizer.sample_sequences(
    num_sequences=667,
    sequence_length=150
)

if isinstance(synthetic_data, dict):
    synthetic_data = synthetic_data["time_series"]

os.makedirs("./data_features/synth", exist_ok=True)
synthetic_data.to_csv("./data_features/synth/synthetic_time_series_100k.csv", index=False)

print(f"Done! Generated {len(synthetic_data)} rows.")
```
